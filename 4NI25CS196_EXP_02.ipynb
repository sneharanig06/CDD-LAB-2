```python
import pandas as pd
import numpy as np
data={
       'Name': ['Amit','Bebo','chetan','Amit','Divya','Ravi'],
        'Age': [25,np.nan,90,56,43,843],
       'Marks':[78,90,np.nan,87,67,45],
       'City':['Mysuru','bengaluru','mysuru',np.nan,'Ranchi','Mumbai']

}
df=pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
print("\nShape of dataset (rows,columns):",df.shape)
print("\nCoumns names:",df.columns.tolist())
print("\nData Types ofeach columns:")
print(df.dtypes)
print("\nDatatyes info;")
df.info()
```

```python
print("\nDescriptive satistics (numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each columns:")
print(df.isnull().sum())
print("\n Total missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicate row in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicates row:",df.duplicated().sum())
print("\nEDA process completed successfully!")
```
